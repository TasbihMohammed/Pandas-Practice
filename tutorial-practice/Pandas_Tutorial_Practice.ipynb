### **Merging DataFrames in Pandas**

Merging is the process of combining two DataFrames based on a common column or key.

It is useful when related information is stored in separate DataFrames and we want to combine it into a single DataFrame.

In this example, we will merge two DataFrames using the `Student_ID` column.

````python
import pandas as pd

# Create the first DataFrame containing student names
students = pd.DataFrame({
    'Student_ID': [1, 2, 3, 4],
    'Name': ['Ali', 'Sara', 'Omar', 'Mona']
})

# Create the second DataFrame containing student grades
grades = pd.DataFrame({
    'Student_ID': [2, 3, 4, 5],
    'Grade': [85, 90, 78, 95]
})

# Display the first DataFrame
print("Students DataFrame:")
print(students)

# Display the second DataFrame
print("\nGrades DataFrame:")
print(grades)
````

### **Inner Merge**

An inner merge returns only the rows that have matching keys in both DataFrames.

Rows that do not have a matching key in the other DataFrame are excluded from the result.

This is the default merge type in Pandas.

````python
# Merge the two DataFrames using the common Student_ID column
# Keep only students whose IDs exist in both DataFrames
inner_merged = pd.merge(
    students,
    grades,
    on='Student_ID',
    how='inner'
)

# Display the merged DataFrame
print(inner_merged)
````

### **Outer Merge**

An outer merge returns all rows from both DataFrames.

If a key exists in one DataFrame but has no matching key in the other DataFrame, the missing values are filled with `NaN`.

This is useful when we want to preserve all records from both DataFrames.

````python
outer_merged=pd.merge(students,grades,how='outer',on='Student_ID')
print(outer_merged)
````

### **Left Merge**

A left merge keeps all rows from the left DataFrame (the first DataFrame).

Matching data from the right DataFrame is added to the result. If there is no matching key, the corresponding values from the right DataFrame are filled with `NaN`.

````python
left_merged= pd.merge(students,grades,how='left',on='Student_ID')
print(left_merged)
````

### **Right Merge**

A right merge keeps all rows from the right DataFrame (the second DataFrame).

Matching data from the left DataFrame is added to the result. If there is no matching key, the corresponding values from the left DataFrame are filled with `NaN`.

````python
right_merged=pd.merge(students,grades,how='right',on='Student_ID')
print(right_merged)
````

### **Merging DataFrames Using Different Column Names**

Sometimes, the columns used to match records have different names in the two DataFrames.

In this case, we can use `left_on` to specify the column name in the left DataFrame and `right_on` to specify the column name in the right DataFrame.

Unlike `on`, which is used when both DataFrames share the same column name, `left_on` and `right_on` allow us to merge DataFrames using differently named columns.

````python
import pandas as pd

# Create the first DataFrame
students = pd.DataFrame({
    'Student_ID': [1, 2, 3, 4],
    'Name': ['Ali', 'Sara', 'Omar', 'Mona']
})

# Create the second DataFrame
grades = pd.DataFrame({
    'ID': [2, 3, 4, 5],
    'Grade': [85, 90, 78, 95]
})

# Merge using columns with different names
merged_data = pd.merge(
    students,
    grades,
    left_on='Student_ID',  # Key column in the left DataFrame
    right_on='ID',         # Key column in the right DataFrame
    how='inner'            # Keep matching records only
)

# Remove the extra ID column after merging
merged_data = merged_data.drop(columns=['ID'])

# Display the cleaned DataFrame
print(merged_data)
````

### **Summary: Concatenation in Pandas**

The `pd.concat()` function combines two or more DataFrames along a specified axis.

- `axis=0`: Adds rows vertically.
- `axis=1`: Adds columns horizontally and aligns rows by their index labels.
- `ignore_index=True`: Creates a new sequential index for the concatenation result.
- `NaN` appears when data is missing after alignment.

**Difference between `concat()` and `merge()`:**

Use `concat()` mainly to stack datasets or place them side by side. Use `merge()` to combine related data based on matching key columns.

````python
import pandas as pd

# Create the first DataFrame
students_1 = pd.DataFrame({
    'Name': ['Ali', 'Sara'],
    'Grade': [85, 90]
})

# Create the second DataFrame
students_2 = pd.DataFrame({
    'Name': ['Omar', 'Mona'],
    'Grade': [78, 95]
})

# Display the first DataFrame
print("Students 1:")
print(students_1)

# Display the second DataFrame
print("\nStudents 2:")
print(students_2)
````

````python
all_students= pd.concat([students_1,students_2],axis=0,ignore_index=True)
print(all_students)
````

### **Detecting Null Values**

Missing values are values that are absent from a dataset.

Pandas provides several methods to detect missing values, such as `isnull()` and `notnull()`.

- `isnull()` returns `True` for missing values and `False` otherwise.
- `notnull()` returns `True` for non-missing values and `False` for missing values.

These methods help us identify missing data before performing analysis.

````python
import pandas as pd
import numpy as np

# Create a DataFrame containing some missing values
students = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Omar', 'Mona','Yara'],
    'Age': [19, np.nan, 20, np.nan ,22],
    'Grade': [85, 90, np.nan, 95,np.nan]
})

# Display the DataFrame
print(students)
````

````python
print(students.isnull())
````

````python
print(students.notnull().sum())
````

````python
# Count all missing values in the entire DataFrame
print(students.isnull().sum().sum())
````

### **Handling Null Values**

After detecting missing values, we need to decide how to handle them.

Pandas provides methods such as `dropna()` to remove missing data and `fillna()` to replace missing values.

The appropriate method depends on the dataset and the purpose of the analysis. Removing data may cause information loss, while filling missing values requires choosing a suitable replacement.

### Removing Missing Values with `dropna()`

The `dropna()` method removes rows containing missing values by default.

It is useful when rows with incomplete data should not be included in the analysis.

By default, `dropna()` returns a new DataFrame and does not modify the original DataFrame in place.

````python
clean_students = students.dropna()
print(clean_students)
````

````python
clean_columns = students.dropna(axis=1)
print(clean_columns)
````

### **Filling Missing Values with `fillna()`**

The `fillna()` method replaces missing values with specified values.

It can be used to fill missing values with a constant, such as `0` or `"Unknown"`, or with calculated values, such as the mean or median of a column.

The replacement strategy should be selected based on the meaning of the data.

````python
# Replace all missing values with 0
filled_students = students.fillna(0)
print(filled_students)
````

````python
mean_grade = students['Grade'].fillna(students['Grade'].mean())
median_age = students['Age'].fillna(students['Age'].interpolate())
students.fillna({'Grade':mean_grade,'Age':median_age},inplace=True)
print(students)
````

````python
#The interpolate() method in Pandas:
# is used to estimate missing values based on
# existing values in a dataset.

#By default, it uses linear interpolation,
# which estimates missing values
# by assuming a linear relationship between the available values.

#It is useful when the values follow a sequence,
# such as time series data or measurements collected over time.

temperature = pd.Series([10, 20, np.nan, 40, 50])
temperature.fillna(temperature.interpolate(),inplace=True)
print(temperature)
````

### **Grouping Data with `groupby()`**

The `groupby()` method in Pandas is used to group rows based on one or more columns.

After grouping the data, we can apply aggregate functions such as `sum()`, `mean()`, `count()`, `min()`, and `max()` to summarize each group.

The grouping process follows three steps:

1. **Split:** Divide the data into groups based on a column or columns.
2. **Apply:** Apply a function to each group.
3. **Combine:** Combine the results into a summarized output.

````python
obesity_data.groupby('Gender').count()
````

````python
obesity_data.groupby('Gender')['Height'].mean()
````

````python
obesity_data.groupby('Gender')['Weight'].aggregate(['mean','median','std'])
````
